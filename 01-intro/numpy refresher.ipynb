## Multi-dimensional arrays

```python
np.zeros((3,4))
```

```python
bar = np.array([
        [1,2,3],
        [4,5,6]
    ])
```

```python
bar[0,2]
```

```python
bar[0,2] = 2 # multi-dim arrays are mutable
```

```python
bar[1] # accessing rows
```

```python
bar[:,1] # accessing columns
```

## Randomly generated arrays

```python
np.random.seed(2)
np.random.rand(5,2)
```

```python
np.random.seed(2)
np.random.randn(5,2) # randn ~ random sample from standard normal dist
```

```python
np.random.seed(2)
np.random.randint(low=0, high=100, size=(5,2))
```
